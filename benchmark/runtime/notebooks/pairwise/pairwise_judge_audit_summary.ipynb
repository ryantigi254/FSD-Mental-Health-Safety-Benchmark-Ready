# Pairwise Judge Audit Summary

Reads canonical judge-audit reports only.

```python
from pathlib import Path
import json

report_root = Path(r'''/Users/ryangichuru/Documents/SSD-K/Uni/3rd year/NLP-ready-pairwise/benchmark/runtime/metric-results/pairwise/judge_audit/reports''')
report_paths = sorted(report_root.glob('*.json'))
print('judge-audit reports found:', len(report_paths))
reports = [json.loads(path.read_text(encoding='utf-8')) for path in report_paths]
```

```python
if reports:
    for report in reports:
        print('\nreport:', report.get('report_name'))
        for judge_id, metrics in report.get('per_judge', {}).items():
            print(judge_id, {key: value for key, value in metrics.items() if key != 'item_summaries'})
```
