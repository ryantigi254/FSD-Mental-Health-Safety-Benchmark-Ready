```python
from pathlib import Path
import json

report_root = Path(r'''/Users/ryangichuru/Documents/SSD-K/Uni/3rd year/NLP-ready-pairwise/benchmark/runtime/metric-results/pairwise/reports''')
report_paths = sorted(report_root.glob('*/*__pairwise_secondary_results.json'))
print('reports found:', len(report_paths))
reports = [json.loads(path.read_text(encoding='utf-8')) for path in report_paths]
```

```python
if reports:
    for report in reports:
        summary = report.get('execution_summary', {})
        print(report.get('slice_id'), report.get('pooled_complete'), report.get('case_manifest_status'), summary.get('routine_two_judge_results', {}), summary.get('escalated_four_judge_results', {}))
```

```python
if reports and 'judge_diagnostics' == 'judge_diagnostics':
    for report in reports:
        print('\n', report.get('slice_id'))
        print(report.get('judge_agreement', {}))
```

```python
if reports and 'judge_diagnostics' == 'stacked_execution':
    for report in reports:
        print('\n', report.get('slice_id'))
        print(report.get('execution_summary', {}))
```

```python
if reports and 'judge_diagnostics' == 'escalation':
    for report in reports:
        summary = report.get('execution_summary', {})
        print('\n', report.get('slice_id'))
        print('escalation_summary:', summary.get('escalation_summary', {}))
        print('persistent_disagreement_cases:', summary.get('persistent_disagreement_cases', [])[:10])
```
